```python
import sys
sys.path.append('../src')

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from metrics import evaluate_and_save

df = pd.read_csv('../data/processed/darija_clean.csv')

X_train, X_test, y_train, y_test = train_test_split(
    df['clean_text'], df['label'], test_size=0.2, random_state=42, stratify=df['label']
)

# Vectorisation TF-IDF (mots + n-grammes de caractères pour capturer la Darija)
vectorizer = TfidfVectorizer(ngram_range=(1, 2), max_features=15000)
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

# Modèle 1 : Logistic Regression
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train_vec, y_train)
evaluate_and_save("Baseline_LogReg", y_test, lr.predict(X_test_vec))

# Modèle 2 : Linear SVM
svm = LinearSVC()
svm.fit(X_train_vec, y_train)
evaluate_and_save("Baseline_LinearSVM", y_test, svm.predict(X_test_vec))
```
