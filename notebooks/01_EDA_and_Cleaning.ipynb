```python
import sys
sys.path.append('../src')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from preprocessing import clean_darija_text

# 1. Chargement du dataset MAC (Moroccan Arabic Corpus) depuis GitHub
url = "https://raw.githubusercontent.com/abdeljalilELmajjodi/MAC-Moroccan-Arabic-Corpus/master/MAC%20corpus.csv"
df = pd.read_csv(url)

# Détection automatique des colonnes texte et label
text_col = df.columns[0]
label_col = df.columns[1]

df = df[[text_col, label_col]].rename(columns={text_col: 'text', label_col: 'label'}).dropna()
df['clean_text'] = df['text'].astype(str).apply(clean_darija_text)
df = df[df['clean_text'].str.len() > 2]

# 2. Sauvegarde
df.to_csv('../data/processed/darija_clean.csv', index=False)
print(f"Dataset propre sauvegardé : {len(df)} lignes.")
print(df.head())

# 3. Graphique EDA
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='label', palette='viridis')
plt.title("Distribution des classes (MAC Darija Dataset)")
plt.savefig('../results/plots/class_distribution.png', bbox_inches='tight')
plt.show()
```
