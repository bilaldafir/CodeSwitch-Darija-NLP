```python
import sys
sys.path.append('../src')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datasets import load_dataset
from preprocessing import clean_darija_text

# 1. Chargement du dataset Darija
dataset = load_dataset("atlasia/darija-sentiment-analysis")
df = pd.DataFrame(dataset['train'])

# Détection automatique des colonnes texte et label
text_col = [c for c in df.columns if 'text' in c.lower() or 'tweet' in c.lower() or 'comment' in c.lower()][0]
label_col = [c for c in df.columns if 'label' in c.lower() or 'sentiment' in c.lower()][0]

df = df[[text_col, label_col]].rename(columns={text_col: 'text', label_col: 'label'})
df['clean_text'] = df['text'].apply(clean_darija_text)
df = df[df['clean_text'].str.len() > 2].dropna()

# 2. Sauvegarde
df.to_csv('../data/processed/darija_clean.csv', index=False)
print(f"Dataset propre sauvegardé : {len(df)} lignes.")
print(df.head())

# 3. Graphique EDA
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='label', palette='viridis')
plt.title("Distribution des classes (Darija)")
plt.savefig('../results/plots/class_distribution.png', bbox_inches='tight')
plt.show()
```
